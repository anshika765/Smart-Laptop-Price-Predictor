In this notebook, the saved preprocessing pipeline and tuned Random Forest model will be loaded and used to predict the price of a new laptop.

The tuned Random Forest model saved during Day 4 is loaded using Joblib.

## 7. Creating a Reusable Prediction Function

A reusable function is created so that laptop specifications can be provided as input and the trained model can directly return the predicted price.

```python
def predict_laptop_price(laptop_data):
    laptop_df = pd.DataFrame([laptop_data])

    processed_data = preprocessor.transform(laptop_df)

    prediction = model.predict(processed_data)

    return prediction[0]
```

```python
laptop = {
    "Company": "HP",
    "TypeName": "Notebook",
    "Inches": 15.6,
    "CPU_Company": "Intel",
    "CPU_Frequency (GHz)": 2.5,
    "RAM (GB)": 8,
    "GPU_Company": "Intel",
    "Weight (kg)": 2.0,
    "SSD_GB": 256,
    "HDD_GB": 0,
    "Flash_GB": 0,
    "Hybrid_GB": 0,
    "Total_Storage_GB": 256,
    "Storage_Type": "SSD",
    "Resolution_Width": 1920,
    "Resolution_Height": 1080,
    "Touchscreen": 0,
    "IPS": 0,
    "CPU_Family": "Core i5",
    "GPU_Family": "Intel HD",
    "OS_Family": "Windows"
}

price = predict_laptop_price(laptop)

print(f"Predicted Price: €{price:,.2f}")
```

## 9. Storing Prediction Results

The predictions from multiple laptop configurations are stored in a DataFrame for easy comparison and analysis.

```python
prediction_results = []

for i, laptop in enumerate(test_laptops, start=1):
    price = predict_laptop_price(laptop)
    
    prediction_results.append({
        "Laptop": f"Laptop {i}",
        "Company": laptop["Company"],
        "RAM (GB)": laptop["RAM (GB)"],
        "CPU_Family": laptop["CPU_Family"],
        "Storage_Type": laptop["Storage_Type"],
        "Predicted_Price_Euro": round(price, 2)
    })

prediction_results_df = pd.DataFrame(prediction_results)

prediction_results_df
```
